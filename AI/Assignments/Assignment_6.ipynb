**Metrics:**
Your task is to implement the following formulas:

**1. Inertia / WCSS:**
The sum of squared distances between each data point and its corresponding cluster centroid.
$$ WCSS = \sum_{i=1}^{k} \sum_{x \in C_i} ||x - \mu_i||^2 $$
where $C_i$ is the $i$-th cluster and $\mu_i$ is its centroid.

**2. Dunn Index:**
The ratio of the minimal inter-cluster distance to the maximal intra-cluster distance.
$$ Dunn = \frac{\min_{1 \le i < j \le k} d(C_i, C_j)}{\max_{1 \le m \le k} diam(C_m)} $$
where $d(C_i, C_j)$ is the distance between cluster $i$ and cluster $j$ (minimum distance between any point in $C_i$ and any point in $C_j$), and $diam(C_m)$ is the maximum distance between any two points within cluster $m$.

### Provided Dataset
We will use the **Iris Dataset**, a classic dataset in machine learning, containing 4 features (sepal length, sepal width, petal length, petal width) for 150 iris flowers.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans

# Load the Iris dataset
iris = load_iris()
X = iris.data  # We will use all 4 features

# Example: Run K-Means with k=3
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
kmeans.fit(X)

labels = kmeans.labels_
centroids = kmeans.cluster_centers_

print("Scikit-learn calculated Inertia:", kmeans.inertia_)
```
